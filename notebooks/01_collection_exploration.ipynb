# 01. Data Collection & Initial Exploration

**Mục tiêu:**
- Kiểm tra đường dẫn và đọc bộ dữ liệu gốc từ `../data/raw/`.
- Xác nhận kích thước (số dòng, số cột), tên cột, kiểu dữ liệu và tỉ lệ giá trị thiếu.
- Ghi nhận các quan sát ban đầu để phục vụ bước làm sạch và chốt 6 câu hỏi nghiên cứu.

```python
from pathlib import Path
import pandas as pd
import numpy as np

RAW_DIR = Path("../data/raw")
print("Raw directory:", RAW_DIR.resolve())
print("Files found:", [p.name for p in RAW_DIR.iterdir() if p.name != ".gitkeep"])
```

```python
DATA_FILE = RAW_DIR / "VietJobs.csv"

if DATA_FILE.exists():
    df = pd.read_csv(DATA_FILE)
    print(f"Shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
    display(df.head())
else:
    print(f"File not found: {DATA_FILE}. Hãy tải dữ liệu vào data/raw/ và cập nhật DATA_FILE.")
```

```python
# Kiểm tra kiểu dữ liệu và tỉ lệ giá trị thiếu
if "df" in globals():
    summary = pd.DataFrame({
        "dtype": df.dtypes,
        "non_null": df.notna().sum(),
        "missing_count": df.isna().sum(),
        "missing_pct": (df.isna().mean() * 100).round(2),
        "n_unique": df.nunique(dropna=True),
    })
    display(summary)
```

## Phân bố các cột phân loại chính

```python
if "df" in globals():
    for col in ["category", "experience_required", "contract_type", "country"]:
        display(df[col].value_counts().head(20).to_frame())
```
